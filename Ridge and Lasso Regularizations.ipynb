```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Lasso,Ridge
from sklearn.metrics import mean_squared_error
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
np.random.seed(42)
x=np.random.rand(100,1)*10
y=3*x**2+2*x+np.random.randn(100,1)*10
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
x_poly=PolynomialFeatures(degree=2)
x_train_poly=x_poly.fit_transform(x_train)
x_test_poly=x_poly.transform(x_test)
y_train_poly=y_train
y_test_poly=y_test
ridge_model=Ridge(alpha=1.0)
ridge_model.fit(x_train_poly,y_train_poly)
ridge_pred=ridge_model.predict(x_test_poly)
Lasso_model=Lasso(alpha=1.0)
Lasso_model.fit(x_train_poly,y_train_poly)
Lasso_pred=Lasso_model.predict(x_test_poly)
ridge_mse=mean_squared_error(y_test_poly,ridge_pred)
Lasso_mse=mean_squared_error(y_test_poly,Lasso_pred)
print("Ridge MSE:",ridge_mse)
print("Lasso MSE:",Lasso_mse)
```
